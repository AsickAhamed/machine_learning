```python
# import necessary libs
import pandas as pd
import numpy as np
```

```python
# loading data
x = np.array([])
y = np.array([])
```

```python
# cost function
def compute_cost(x,y,w,b):

  f_x = (w * x) + b
  cost = np.sum((f_x - y) ** 2)
  return cost / (2 * len(x))
```

```python
def compute_gradient(x,y,w,b):

  m = len(x)
  f_x = (w * x) + b
  dj_dw = ((f_x - y) * x) / m
  dj_db = (f_x - y) / m

  return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w, b, alpha, num_iters):

  m = x.shape[0]
  print(f"m: {m}")
  cost = []
  p_hist = []
  for i in range(1,num_iters+1):
    dj_dw, dj_db = compute_gradient(x, y, w, b)
    w = w - alpha * dj_dw
    b = b - alpha * dj_db

    j = compute_cost(x, y, w, b)
    cost.append(j)
    p_hist.append([w, b])

    if i % 100 == 0:
      print(f"Iteration {i}: Cost {j} ",
            f"dj_dw: {dj_dw}, dj_db: {dj_db}")


  return w, b



```

```python
%pdb on
x = np.array([1.0, 2.0])
y = np.array([300.0, 500.0])
w = np.array([0])
b = np.array([0])
alpha = 0.1
iters = 1000
w, b = gradient_descent(x,y, w, b, alpha, iters)
print(f"w: {w} \nb : {b}")
estimation = (w * x[0]) + b
print(f"estimation: {estimation} || actual : {y[0]}")
%pdb off
```
